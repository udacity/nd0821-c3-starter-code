## Cleanup Data

```python
from pathlib import Path

import pandas as pd


# base_dir = Path(__file__).resolve().parent
# csv_path = base_dir / "census.csv"
# out_path = base_dir / "census_data_new.csv"

csv_path = "census_old.csv"
out_path = "census.csv"

# Open the CSV in pandas so the whitespace issue can be observed.
df = pd.read_csv(csv_path)
print("Original columns:")
print(df.columns.tolist())
print("\nPreview:")
print(df.head())
```

```python

# Remove leading/trailing spaces from column names and string values without
# renaming columns to satisfy Python naming conventions.
df.columns = df.columns.str.strip()
df = df.apply(lambda col: col.map(lambda x: x.strip() if isinstance(x, str) else x))

# Save the cleaned data to a new CSV in the same directory.
df.to_csv(out_path, index=False)
print(f"\nSaved cleaned census data to: {out_path}")
```
